## Sel 2 — Tampilkan website
Gambar dari folder `assets/` diubah menjadi base64 agar tampil di dalam Colab.

```python
import base64, re, os
from IPython.display import HTML, display, IFrame

BASE = "/content/site"
if not os.path.exists(f"{BASE}/index.html"):
    # cari index.html di subfolder
    for r, d, f in os.walk(BASE):
        if "index.html" in f:
            BASE = r; break

html = open(f"{BASE}/index.html", encoding="utf-8").read()

def inline(m):
    path = os.path.join(BASE, m.group(1))
    if not os.path.exists(path):
        return m.group(0)
    mime = 'png' if path.endswith('.png') else 'jpeg'
    data = base64.b64encode(open(path, "rb").read()).decode()
    return f'src="data:image/{mime};base64,{data}"'

html_inline = re.sub(r'src="(assets/[^"]+)"', inline, html)
open("/content/preview.html", "w", encoding="utf-8").write(html_inline)

display(HTML(f'<iframe srcdoc="{html_inline.replace(chr(34), "&quot;")}" width="100%" height="900" style="border:1px solid #ccc;border-radius:8px"></iframe>'))
```
